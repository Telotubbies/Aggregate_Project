# 01 — Model Training: aggregate instance segmentation

Baseline + ablation pipeline for `yolo26n-seg` on the cleaned datasets from `00EDA.ipynb`.

**Experiment matrix (minimal-first, approved):**

| phase | what | why |
|---|---|---|
| 0 | environment + dataset wiring | reproducibility audit |
| 1 | smoke test (overfit 24 imgs) | prove pipeline learns before burning GPU-hours |
| 2 | autobatch probe | batch=-1 -> max safe batch on this GPU |
| 3 | optimizer bake-off: MuSGD vs AdamW vs SGD | official YOLO26 recipe uses MuSGD; decide empirically |
| 4 | data arms: `gold` vs `gold_aug` | does offline expansion (incl. dark/rain) actually help |
| 5 | holdout `test` — evaluated **once**, on the winner only | untouched final claim |

**Auto rules wired in:**
- early stop: `patience` on val fitness (0.1*mAP50 + 0.9*mAP50-95, mask)
- auto-LR retry: if a run diverges or plateaus at a clearly-broken level -> retry once at lr0/2 and keep the better run
- auto-batch: `check_train_batch_size` probe before any real run

**Periodic viz:** a fixed panel of validation images is predicted every `VIZ_EVERY`
epochs during each real run (callback -> PNG under `runs/<name>/viz/`).

Requirements: `ultralytics>=8.4.146`, torch with CUDA/ROCm (`pip install ultralytics`).
Datasets expected at `data/gold`, `data/gold_aug` (optionally `data/gold_clahe`) —
on a fresh machine either copy `experiment/data/` from the release asset or re-run `00EDA.ipynb`.

## 1. Shared machinery

Helpers every phase uses: aug policy from `train.yaml`, fixed val panel,
periodic-viz callback, results parsing, and `train_with_retry` (auto-LR).

```python
# ============================================================
# 1.1 HELPERS — aug policy, results parsing, fitness
# ============================================================

# ---- aug policy: single source of truth is train.yaml --------
AUG_KEYS = ["mosaic", "close_mosaic", "copy_paste", "fliplr", "flipud",
            "degrees", "scale", "translate", "perspective", "shear",
            "hsv_h", "hsv_s", "hsv_v", "erasing"]
_train_yaml = yaml.safe_load((EXP_DIR / "train.yaml").read_text())
AUG = {k: _train_yaml[k] for k in AUG_KEYS if k in _train_yaml}
print("aug policy:", AUG)


def read_results(run_dir):
    """ultralytics results.csv -> DataFrame (column names stripped)."""
    df = pd.read_csv(Path(run_dir) / "results.csv")
    df.columns = [c.strip() for c in df.columns]
    return df


def map_col(df):
    """mask mAP50-95 column name, version-tolerant."""
    for c in df.columns:
        if "map50-95(m" in c.lower():
            return c
    for c in df.columns:
        if "map50-95" in c.lower():
            return c
    raise KeyError(f"no mAP50-95 col in {list(df.columns)}")


def run_fitness(run_dir):
    """fitness = 0.1*mask mAP50 + 0.9*mask mAP50-95 (ultralytics convention)."""
    df = read_results(run_dir)
    m50  = df[[c for c in df.columns if "map50(" in c.lower() and "95" not in c.lower()][-1]]
    m95  = df[map_col(df)]
    fit  = 0.1 * m50 + 0.9 * m95
    return df, fit
```

```python
# ============================================================
# 1.2 FIXED VIZ PANEL + PERIODIC-VIZ CALLBACK + TRAIN WRAPPER
# ============================================================

# ---- panel: 6 fixed valid images — flagged-first if flags exist
val_imgs = sorted((DATASETS["gold"] / "valid" / "images").iterdir())
qf_path  = DATASETS["gold"] / "quality_flags.csv"
panel = []
if qf_path.exists():
    qf = pd.read_csv(qf_path)
    qv = qf[qf.get("split", pd.Series(["valid"]*len(qf))).astype(str).str.contains("valid")] \
           if "split" in qf.columns else qf
    flag_cols = [c for c in qf.columns
                 if ("flag" in c.lower() or c.startswith("is_"))
                 and qf[c].dtype != object]
    if flag_cols:
        flagged = qv[qv[flag_cols].any(axis=1)]["image"].tolist() if "image" in qv.columns else []
        panel = flagged[:3]
panel += [p.name for p in val_imgs if p.name not in panel][:6 - len(panel)]
PANEL = [str(DATASETS["gold"] / "valid" / "images" / n) for n in panel]
print("viz panel:", [Path(p).name[:40] for p in PANEL])

VIZ_EVERY = 10

# ---- thesis metric alignment --------------------------------------------------
# ultralytics SegmentMetrics.fitness returns BOX fitness -> early-stop AND
# best.pt selection would be driven by box mAP, not the mask metric we report.
# Patch to mask fitness so checkpoints/stops align with the thesis claim.
from ultralytics.utils.metrics import SegmentMetrics
SegmentMetrics.fitness = property(lambda m: m.seg.fitness())
print("patched: SegmentMetrics.fitness -> mask (seg) fitness")

def viz_cb(trainer):
    """every VIZ_EVERY epochs (and last): predict panel -> runs/<name>/viz/."""
    ep = trainer.epoch + 1
    if ep % VIZ_EVERY and ep != trainer.epochs:
        return
    try:
        out = Path(trainer.save_dir) / "viz"
        out.mkdir(exist_ok=True)
        y = YOLO(str(trainer.last))
        rs = y.predict(PANEL, imgsz=640, device=trainer.device, verbose=False)
        fig, axes = plt.subplots(1, len(rs), figsize=(3 * len(rs), 3))
        for ax, r in zip(np.atleast_1d(axes), rs):
            ax.imshow(r.plot()[..., ::-1]); ax.axis("off")
        fig.savefig(out / f"ep{ep:03d}.png", dpi=90, bbox_inches="tight")
        plt.close(fig)
    except Exception as e:                       # viz must never kill training
        print(f"[viz_cb] ep{ep} skipped: {e}")


def ds_sig(data_yaml):
    """cheap dataset signature: file count + total MB — pinned in every rec."""
    root = Path(yaml.safe_load(Path(data_yaml).read_text())["path"])
    n = sz = 0
    for p in root.rglob("*"):
        if p.is_file():
            n += 1; sz += p.stat().st_size
    return f"{n}f:{sz // 2**20}MB"

import subprocess
def _git_sha():
    try:
        return subprocess.run(["git", "rev-parse", "--short", "HEAD"],
                              capture_output=True, text=True, cwd=EXP_DIR,
                              timeout=5).stdout.strip() or "nogit"
    except Exception:
        return "nogit"
GIT_SHA = _git_sha()


def train_run(name, data, model_src="yolo26n-seg.pt", epochs=100, patience=30,
              batch=16, optimizer="AdamW", lr0=1e-3, lrf=0.01, momentum=0.937,
              weight_decay=5e-4, warmup_epochs=3.0, viz=True, seed=SEED, **kw):
    """one training run. returns dict(record)."""
    t0 = time.time()
    np.random.seed(seed); torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)
    model = YOLO(model_src)
    if viz:
        model.add_callback("on_train_epoch_end", viz_cb)
    args = dict(data=data, task="segment", epochs=epochs, patience=patience,
                batch=batch, imgsz=640, device=DEVICE, workers=WORKERS,
                optimizer=optimizer, lr0=lr0, lrf=lrf, momentum=momentum,
                weight_decay=weight_decay, warmup_epochs=warmup_epochs,
                cos_lr=True, amp=True, seed=seed, deterministic=True,
                project=str(RUNS_DIR), name=name, exist_ok=False,
                plots=True, save_period=-1, verbose=False,
                **{**AUG, **kw})
    model.train(**args)
    save_dir = Path(model.trainer.save_dir)
    df, fit = run_fitness(save_dir)
    rec = {"name": name, "data": Path(str(data)).stem, "model": str(model_src),
           "optimizer": optimizer, "lr0": lr0, "batch": batch, "seed": seed,
           "ds_sig": ds_sig(data), "git": GIT_SHA,
           "epochs_run": len(df), "best_fit": float(fit.max()),
           "best_epoch": int(fit.idxmax()) + 1,
           "final_fit": float(fit.iloc[-1]),
           "wall_min": round((time.time() - t0) / 60, 1),
           "save_dir": str(save_dir)}
    return rec


def needs_lr_retry(rec):
    """auto-LR trigger: diverged (fit ~0 / NaN) or hard-stalled low."""
    if not np.isfinite(rec["best_fit"]) or rec["best_fit"] < 0.02:
        return True
    df, fit = run_fitness(rec["save_dir"])
    tail = fit.iloc[max(0, len(fit) - 10):]      # last <=10 epochs
    return bool(len(fit) >= 10 and rec["best_fit"] < 0.15
                and (tail.max() - tail.min()) < 0.005)


def train_with_retry(name, **kw):
    """train; if needs_lr_retry -> one retry at lr0/2, keep the better."""
    rec = train_run(name, **kw)
    if needs_lr_retry(rec) and kw.get("lr0", 1e-3) > 1e-5:
        print(f"[{name}] weak run (best_fit={rec['best_fit']:.3f}) "
              f"-> retry lr0 {kw['lr0']} -> {kw['lr0']/2:g}")
        rec2 = train_run(name + "_lrhalf", **{**kw, "lr0": kw["lr0"] / 2})
        if rec2["best_fit"] > rec["best_fit"]:
            rec = rec2
    return rec
```

## 4. Optimizer bake-off — MuSGD vs AdamW vs SGD

Reference points (official Ultralytics YOLO26 recipe, docs/en/guides/yolo26-training-recipe):
- COCO-trained YOLO26 uses **MuSGD**; n-size recipe: `lr0=0.0054, momentum=0.947, wd=0.00064, warmup~1ep`
- `optimizer='auto'` picks MuSGD only above ~10k iterations — our dataset is small
  enough that auto would silently pick AdamW, so we decide explicitly.
- AdamW fine-tune convention: `lr0=1e-3`; SGD convention: `lr0=1e-2`.

Protocol: 30 epochs on `gold_aug` (the strongest candidate data arm), identical
seed/batch/aug, `patience=10`, `cos_lr=True`. Auto-LR retry still armed.
Winner = highest best mask fitness; tie -> fewer epochs to best.

**Disclosed caveats (goes in the thesis limitations):**
- this is a *screening* bake-off at 30 epochs — ranking can flip at 100 epochs
  (Ultralytics tuning docs warn short-run hyperparameters rarely transfer);
- the auto-LR rule (diverged fit<0.02, or stalled-low fit<0.15 with <0.005
  wiggle over the last 10 epochs -> retry at lr0/2 once) is an ad-hoc guard,
  not a published method — every retry is logged as a separate run in
  `optimizer_bakeoff.csv` for full disclosure;
- optimizer is chosen on `gold_aug` and reused for `gold` — a confound we
  accept for budget; noted for the write-up.

```python
# ============================================================
# 4.1 BAKE-OFF — 3 optimizers x 30 epochs on gold_aug
# ============================================================
# note: ultralytics maps `momentum` -> Adam beta1 for Adam-family; use std 0.9
OPT_GRID = [
    dict(optimizer="MuSGD", lr0=5.4e-3, momentum=0.947, weight_decay=6.4e-4, warmup_epochs=1.0),
    dict(optimizer="AdamW", lr0=1.0e-3, momentum=0.9,   weight_decay=5.0e-4, warmup_epochs=3.0),
    dict(optimizer="SGD",   lr0=1.0e-2, momentum=0.937, weight_decay=5.0e-4, warmup_epochs=3.0),
]

opt_recs = []
for cfg in OPT_GRID:
    rec = train_with_retry(f"opt_{cfg['optimizer']}",
                           data=DATA_YAML["gold_aug"],
                           epochs=30, patience=10, batch=BATCH,
                           lrf=0.01, viz=True, **cfg)
    opt_recs.append(rec)
    print(rec)

opt_df = pd.DataFrame(opt_recs)
opt_df.to_csv(RES_DIR / "optimizer_bakeoff.csv", index=False)
opt_df
```

```python
# ============================================================
# 4.2 VIZ — per-optimizer fitness curves + loss curves
# ============================================================
fig, ax = plt.subplots(1, 2, figsize=(11, 3.6))
for rec in opt_recs:
    df, fit = run_fitness(rec["save_dir"])
    lbl = f"{rec['optimizer']} (lr0={rec['lr0']:g})"
    ax[0].plot(df["epoch"], fit, label=lbl)
    if "val/seg_loss" in df.columns:
        ax[1].plot(df["epoch"], df["val/seg_loss"], label=lbl)
ax[0].set_title("mask fitness / epoch"); ax[0].legend(fontsize=8)
ax[1].set_title("val seg_loss / epoch"); ax[1].legend(fontsize=8)
plt.tight_layout(); plt.show(); plt.close("all")

print(opt_df[["name","optimizer","lr0","best_fit","best_epoch","wall_min"]]
      .sort_values("best_fit", ascending=False).to_string(index=False))
```

### 4.3 DECISION — winner optimizer

Selection is automatic but logged: max `best_fit`; ties broken by earlier
`best_epoch` (faster convergence). Whatever wins is frozen into `WIN_OPT`
for the data-arm phase — do not hand-edit after looking at val.

```python
best_row = opt_df.sort_values(["best_fit", "best_epoch"],
                              ascending=[False, True]).iloc[0]
WIN_OPT = next(c for c in OPT_GRID if c["optimizer"] == best_row["optimizer"])
WIN_OPT = dict(WIN_OPT, lr0=float(best_row["lr0"]))   # keep retried lr if it won
print("WIN_OPT =", WIN_OPT)
```

## 5. Data arms — `gold` vs `gold_aug`

Does offline expansion (4x train, incl. dark/rain env variants) actually beat
raw gold? Same winner optimizer, same batch/seed, `epochs=100, patience=30`.
Periodic panel-viz PNGs land under `runs/arm_*/viz/` every 10 epochs.

```python
# ============================================================
# 5.1 ARM RUNS — winner optimizer on gold and gold_aug
# ============================================================
ARMS = ["gold", "gold_aug"]
if "gold_clahe" in DATA_YAML:
    ARMS.append("gold_clahe")

SEEDS = [SEED]          # minimal-first default; e.g. [42, 7, 123] for
                        # 3-seed replication -> mean +/- std reporting
arm_recs = []
for sd in SEEDS:
    for arm in ARMS:
        run_name = f"arm_{arm}_s{sd}" if len(SEEDS) > 1 else f"arm_{arm}"
        rec = train_with_retry(run_name, data=DATA_YAML[arm],
                               epochs=100, patience=30, batch=BATCH,
                               lrf=0.01, viz=True, seed=sd, **WIN_OPT)
        rec["arm"] = arm
        arm_recs.append(rec)
        print(rec)

arm_df = pd.DataFrame(arm_recs)
arm_df.to_csv(RES_DIR / "data_arms.csv", index=False)
arm_df
```

```python
# ============================================================
# 5.2 VIZ — fitness curves, per-class mask AP, pred panels
# ============================================================
fig, ax = plt.subplots(1, 2, figsize=(11, 3.8))
per_class = {}
for rec in arm_recs:
    df, fit = run_fitness(rec["save_dir"])
    ax[0].plot(df["epoch"], fit, label=rec["name"])
    bm = YOLO(Path(rec["save_dir"]) / "weights" / "best.pt")
    m = bm.val(data=DATA_YAML[rec.get("arm", rec["name"].replace("arm_", ""))],
               split="val", imgsz=640, device=DEVICE, verbose=False, plots=False)
    try:
        per_class[rec["name"]] = list(m.seg.maps)          # per-class mAP50-95
    except Exception:
        per_class[rec["name"]] = [float(m.seg.map)] * 2

ax[0].set_title("mask fitness / epoch"); ax[0].legend(fontsize=8)
names = ["gravel", "sand"]
x = np.arange(len(names)); w = 0.8 / len(arm_recs)
for j, (arm, aps) in enumerate(per_class.items()):
    ax[1].bar(x + j * w, aps, w, label=arm)
ax[1].set_xticks(x + w * (len(arm_recs) - 1) / 2); ax[1].set_xticklabels(names)
ax[1].set_title("per-class mask mAP50-95 (val)"); ax[1].legend(fontsize=8)
plt.tight_layout(); plt.show(); plt.close("all")

print(arm_df[["name","optimizer","lr0","best_fit","best_epoch","wall_min"]]
      .sort_values("best_fit", ascending=False).to_string(index=False))
```

```python
# ============================================================
# 5.3 VIZ — best.pt predictions on the fixed panel, per arm
# ============================================================
for rec in arm_recs:
    bm = YOLO(Path(rec["save_dir"]) / "weights" / "best.pt")
    rs = bm.predict(PANEL, imgsz=640, device=DEVICE, verbose=False)
    fig, axes = plt.subplots(1, len(rs), figsize=(3*len(rs), 3))
    for ax_, r in zip(np.atleast_1d(axes), rs):
        ax_.imshow(r.plot()[..., ::-1]); ax_.axis("off")
    fig.suptitle(f"{rec['name']} — fixed panel preds", fontsize=11)
    plt.show(); plt.close("all")
```

### 5.4 DECISION — winning data arm

Winner = highest `best_fit` on **valid**. The loser stays on disk for the
appendix; the winner alone proceeds to the holdout cell.

```python
# winner = arm with highest MEAN best_fit across seeds (1 seed -> same as max)
if "arm" not in arm_df.columns:
    arm_df["arm"] = arm_df["name"].str.replace("arm_", "", regex=False)
score = arm_df.groupby("arm")["best_fit"].agg(["mean", "std", "count"])
print(score.sort_values("mean", ascending=False).to_string())
WIN_ARM = score["mean"].idxmax()
win_arm = arm_df[arm_df["arm"] == WIN_ARM].sort_values(
    "best_fit", ascending=False).iloc[0]
BEST_PT = Path(win_arm["save_dir"]) / "weights" / "best.pt"
print(f"winner: {WIN_ARM}  mean best_fit={score.loc[WIN_ARM,'mean']:.4f}  ckpt={BEST_PT}")
```

## 6. Holdout evaluation — runs exactly once

This is the only cell that touches `test/`. Run it once, after every

### 6.4 FP / FN overlay — per-instance error analysis

Every `test` image overlaid: **green** = TP (mask-IoU >= 0.5, same class),
**red** = FP (predicted instance with no GT), **yellow** = FN (GT never matched).
All overlays saved under `runs/<winner>/fpfn_test/` + `fpfn_stats.csv`;

```python
# ============================================================
# 6.4 FP / FN OVERLAY — greedy mask-IoU matching @0.5 (class-aware)
# ============================================================
import cv2

IOU_TH, CONF_TH = 0.5, 0.25
FPFN_DIR = Path(win_arm["save_dir"]) / "fpfn_test"; FPFN_DIR.mkdir(exist_ok=True)

def load_gt(img_p, lab_dir):
    """YOLO seg labels -> list[rasterized masks], list[cls] at image size."""
    img = cv2.imread(str(img_p)); H, W = img.shape[:2]
    ms, cs = [], []
    lab = lab_dir / (img_p.stem + ".txt")
    if lab.exists():
        for ln in lab.read_text().splitlines():
            t = ln.split()
            if len(t) > 5:
                xy = np.array(t[1:], dtype=np.float32).reshape(-1, 2) * np.array([W, H])
                m = np.zeros((H, W), np.uint8)
                cv2.fillPoly(m, [xy.astype(np.int32)], 1)
                ms.append(m); cs.append(int(float(t[0])))
    return img, ms, cs

rows = []
for ip in sorted((DATASETS[WIN_ARM] / "test" / "images").iterdir()):
    img, gms, gcs = load_gt(ip, DATASETS[WIN_ARM] / "test" / "labels")
    H, W = img.shape[:2]
    r = best_model.predict(str(ip), imgsz=640, conf=CONF_TH,
                           device=DEVICE, verbose=False)[0]
    # masks.xy are polygons in ORIGINAL image coords -> rasterize at (W,H)
    pms = []
    for xy in (r.masks.xy if r.masks is not None else []):
        m = np.zeros((H, W), np.uint8)
        cv2.fillPoly(m, [np.asarray(xy, dtype=np.int32)], 1)
        pms.append(m)
    pcs = [int(c) for c in (r.boxes.cls.cpu().numpy() if r.boxes is not None else [])]
    P, G = len(pms), len(gms)

    iou = np.zeros((P, G))
    for i in range(P):
        for j in range(G):
            if pcs[i] == gcs[j]:
                inter = int((pms[i] & gms[j]).sum())
                iou[i, j] = inter / (pms[i] | gms[j]).sum()
    mp, mg = set(), set()
    for i, j in sorted(zip(*np.where(iou >= IOU_TH)), key=lambda ij: -iou[ij]):
        if i not in mp and j not in mg:
            mp.add(int(i)); mg.add(int(j))
    tp, fp, fn = len(mp), P - len(mp), G - len(mg)

    ov = img.copy().astype(np.float32)
    for j in range(G):                       # green = matched GT / yellow = FN
        c = np.array([0, 200, 0]) if j in mg else np.array([0, 230, 255])
        ov[gms[j] > 0] = 0.55 * ov[gms[j] > 0] + 0.45 * c
    for i in range(P):                       # red = FP pred
        if i not in mp:
            ov[pms[i] > 0] = 0.55 * ov[pms[i] > 0] + 0.45 * np.array([0, 0, 255])
    ov = ov.astype(np.uint8)
    cv2.putText(ov, f"TP{tp} FP{fp} FN{fn}", (8, 24),
                cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 255), 2)
    cv2.imwrite(str(FPFN_DIR / ip.name), ov)
    rows.append(dict(image=ip.name, tp=tp, fp=fp, fn=fn,
                     fp_sand=sum(1 for i in range(P) if i not in mp and pcs[i] == 1),
                     fn_sand=sum(1 for j in range(G) if j not in mg and gcs[j] == 1)))

st = pd.DataFrame(rows)
st.to_csv(FPFN_DIR / "fpfn_stats.csv", index=False)
print(f"test FP/FN @IoU0.5, conf>={CONF_TH}:  TP={st.tp.sum()}  FP={st.fp.sum()}  FN={st.fn.sum()}")
print(f"  sand-only:  FP={st.fp_sand.sum()}  FN={st.fn_sand.sum()}")
print(f"overlays -> {FPFN_DIR}  ({len(st)} images)")

worst = st.nlargest(8, "fn")
fig, axes = plt.subplots(2, 4, figsize=(16, 8))
for ax, (_, rw) in zip(axes.flat, worst.iterrows()):
    ax.imshow(cv2.imread(str(FPFN_DIR / rw["image"]))[..., ::-1])
    ax.set_title(f"{rw['image'][:26]}\nFN={rw.fn} FP={rw.fp}", fontsize=8)
    ax.axis("off")
fig.suptitle("worst-FN test images — green=TP, red=FP, yellow=FN (missed GT)")
plt.tight_layout(); plt.show(); plt.close("all")
```

## 8. P2 ablation — stride-4 detect head

Same recipe as the arm runs (winner optimizer, `gold`, seed 42) but the head
detects on **P2+P3+P4+P5** (`Segment26P2`, proto stays on P3-P5 → stride-4,
no validator patch needed). Weights warm-start from `yolo26n-seg.pt`
(partial transfer — shared layers only).

```python
# ============================================================
# 8.1 Segment26P2 — patch parse_model BEFORE any YOLO() build
# ============================================================
from ultralytics.nn.modules.head import Segment26 as _Segment26, Proto26, Detect
from ultralytics.nn.tasks import parse_model


class Segment26P2(_Segment26):
    """Segment26 with proto consuming x[1:] -> detect P2..P5, proto P3..P5."""

    def __init__(self, nc=80, nm=32, npr=256, reg_max=16, end2end=False, ch=()):
        super().__init__(nc=nc, nm=nm, npr=npr, reg_max=reg_max,
                         end2end=end2end, ch=ch)
        self.proto = Proto26(tuple(ch[1:]), self.npr, self.nm, nc)

    def forward(self, x):
        outputs = Detect.forward(self, x)
        preds = outputs[1] if isinstance(outputs, tuple) else outputs
        proto = self.proto(x[1:])
        if isinstance(preds, dict):
            if "one2one" in preds:
                preds["one2many"]["proto"] = proto
                preds["one2one"]["proto"] = (tuple(p.detach() for p in proto)
                                             if isinstance(proto, tuple)
                                             else proto.detach())
            else:
                preds["proto"] = proto
        if self.training:
            return preds
        return (outputs, proto) if self.export else ((outputs[0], proto), preds)


parse_model.__globals__["Segment26"] = Segment26P2
print("patched: parse_model Segment26 -> Segment26P2")

# ---- arch check before burning GPU: head type + proto stride-4 ----
P2_YAML = str(EXP_DIR / "yolo26n-seg-p2.yaml")
_m = YOLO(P2_YAML)
_h = _m.model.model[-1]
assert isinstance(_h, Segment26P2), f"head is {type(_h)}"

_proto = {}
def _hk(mod, inp, out):
    _proto["in"] = [tuple(f.shape) for f in inp[0]]
    o = out[0] if isinstance(out, tuple) else out
    _proto["out"] = tuple(o.shape)
hh = _h.proto.register_forward_hook(_hk)
_m.model.eval()
with torch.no_grad():
    _m.model(torch.zeros(1, 3, 640, 640,
                         device=next(_m.model.parameters()).device))
hh.remove()
print("proto in :", _proto["in"], "-> out:", _proto["out"])
assert _proto["out"][-2:] == (160, 160), "proto must stay stride-4"
assert len(_proto["in"]) == 3, "proto must consume P3-P5 only"
del _m, _h
if HAS_GPU:
    torch.cuda.empty_cache()
print("P2 arch check OK")
```

```python
## 8.4 P2 warm-transfer ablation — RCA fix
# RCA: cold P2 transferred only 34% (index shift → head/proto/neck random).
# This arm remaps stock yolo26n-seg.pt weights onto the P2 graph
# (backbone 1:1, head level k→k-1, proto 1:1, neck by shape-signature) → ~75% warm.

import re, torch
from ultralytics import YOLO
from ultralytics.nn.tasks import parse_model
from ultralytics.nn.modules.head import Segment26 as _S26, Proto26, Detect
from ultralytics.utils.metrics import SegmentMetrics
from ultralytics.engine.trainer import BaseTrainer
from ultralytics.engine.validator import BaseValidator
from pathlib import Path

WORK = Path.cwd()
RESULTS = WORK / "results"; RESULTS.mkdir(exist_ok=True)
DATA = {"gold": "data/gold/data.yaml"}
IMGSZ, SEED = 640, 42
BATCH, DEVICE, WORKERS, PATIENCE = 12, 0, 8, 30
WINNER_OPT, WINNER_LR = "AdamW", 1e-3
AUG = dict(hsv_h=0.005, hsv_s=0.25, hsv_v=0.20, degrees=0.0, translate=0.02,
           scale=0.10, shear=0.0, perspective=0.0, flipud=0.0, fliplr=0.0,
           mosaic=0.30, mixup=0.0, copy_paste=0.0, erasing=0.0)
SegmentMetrics.fitness = property(lambda m: m.seg.fitness())   # thesis metric alignment

class Segment26P2(_S26):
    def __init__(self, nc=80, nm=32, npr=256, reg_max=16, end2end=False, ch=()):
        super().__init__(nc=nc, nm=nm, npr=npr, reg_max=reg_max, end2end=end2end, ch=ch)
        self.proto = Proto26(tuple(ch[1:]), self.npr, self.nm, nc)
    def forward(self, x):
        outputs = Detect.forward(self, x)
        preds = outputs[1] if isinstance(outputs, tuple) else outputs
        proto = self.proto(x[1:])
        if isinstance(preds, dict):
            if "one2one" in preds:
                preds["one2many"]["proto"] = proto
                preds["one2one"]["proto"] = tuple(p.detach() for p in proto) if isinstance(proto, tuple) else proto.detach()
            else: preds["proto"] = proto
        if self.training: return preds
        return (outputs, proto) if self.export else ((outputs[0], proto), preds)
parse_model.__globals__["Segment26"] = Segment26P2

def build_warm_sd(stock_sd, p2_sd):
    warm = {}
    def sig(sd, i): return tuple(tuple(v.shape) for k, v in sd.items() if re.match(rf"model\.{i}\.", k))
    def lmap(sd, i): return {re.sub(rf"^model\.{i}\.", "", k): v for k, v in sd.items() if re.match(rf"model\.{i}\.", k)}
    for k in p2_sd:  # backbone 1:1
        i = int(re.match(r"model\.(\d+)\.", k).group(1))
        if i <= 10 and k in stock_sd and stock_sd[k].shape == p2_sd[k].shape:
            warm[k] = stock_sd[k]
    for k in p2_sd:  # head 29 <- stock 23
        m = re.match(r"model\.29\.(.*)", k)
        if not m: continue
        suf = m.group(1); cand = f"model.23.{suf}"
        if cand in stock_sd and stock_sd[cand].shape == p2_sd[k].shape:
            warm[k] = stock_sd[cand]; continue
        m2 = re.match(r"(cv\d+|one2one_\w+)\.(\d+)\.(.*)", suf)
        if m2 and int(m2.group(2)) >= 1:
            cand = f"model.23.{m2.group(1)}.{int(m2.group(2))-1}.{m2.group(3)}"
            if cand in stock_sd and stock_sd[cand].shape == p2_sd[k].shape:
                warm[k] = stock_sd[cand]
    stock_list = [i for i in sorted({int(re.match(r"model\.(\d+)\.", k).group(1)) for k in stock_sd}) if 11 <= i <= 23]
    sigs = {i: sig(stock_sd, i) for i in stock_list}; used = set()
    for pi in range(11, 29):  # neck by signature
        ps = sig(p2_sd, pi)
        if not ps: continue
        for sj in stock_list:
            if sj in used or sigs[sj] != ps: continue
            sm, pm = lmap(stock_sd, sj), lmap(p2_sd, pi)
            if sm.keys() == pm.keys():
                for leaf, v in sm.items(): warm[f"model.{pi}.{leaf}"] = v
                used.add(sj)
            break
    return warm

stock_sd = torch.load("yolo26n-seg.pt", map_location="cpu", weights_only=False)["model"].float().state_dict()
m = YOLO("yolo26n-seg-p2.yaml")
warm_sd = build_warm_sd(stock_sd, m.model.state_dict())
missing, unexpected = m.model.load_state_dict(warm_sd, strict=False)
print(f"warm-transfer: {len(warm_sd)}/{m.model.state_dict().__len__()} tensors ({len(warm_sd)/len(m.model.state_dict())*100:.0f}%), missing={len(missing)}")

r = m.train(data=DATA["gold"], task="segment", epochs=100, imgsz=IMGSZ, batch=BATCH,
            device=DEVICE, workers=WORKERS, seed=SEED, project="runs", name="arm_gold_p2w",
            exist_ok=True, pretrained=False, optimizer=WINNER_OPT, lr0=WINNER_LR, lrf=0.01,
            weight_decay=5e-4, momentum=0.937, amp=True, close_mosaic=8, patience=PATIENCE,
            cache=False, save=True, save_period=10, plots=True, val=True, deterministic=True, **AUG)
run_dir = Path(m.trainer.save_dir)
print("run:", run_dir)

best = YOLO(str(run_dir / "weights" / "best.pt"))
mv = best.val(data=DATA["gold"], split="val", imgsz=IMGSZ, batch=BATCH, device=DEVICE, plots=False)
def mask_metrics(mm, tag, split):
    return dict(run=tag, split=split, box_map50=mm.box.map50, box_map=mm.box.map,
                seg_map50=mm.seg.map50, seg_map=mm.seg.map, seg_p=mm.seg.mp, seg_r=mm.seg.mr,
                gravel_map=mm.seg.maps[mm.names.get("gravel", 0)] if "gravel" in mm.names else float("nan"),
                sand_map=mm.seg.maps[mm.names.get("sand", 1)] if "sand" in mm.names else float("nan"))
mt = best.val(data=DATA["gold"], split="test", imgsz=IMGSZ, batch=BATCH, device=DEVICE, plots=False)
rows = [mask_metrics(mv, "p2w", "val"), mask_metrics(mt, "p2w", "test")]
import pandas as pd
prev = pd.read_csv(RESULTS / "metrics_summary.csv")
out = pd.concat([prev, pd.DataFrame(rows)], ignore_index=True)
out.to_csv(RESULTS / "metrics_summary.csv", index=False)
pd.set_option("display.float_format", lambda v: f"{v:.4f}")
show = out[out["run"].isin(["p3-baseline", "p2", "p2w"])]
print(show[["run","split","seg_map50","seg_map","gravel_map","sand_map","seg_p","seg_r"]].to_string(index=False))
```

## 9. Convention-aware evaluation + clean960 arm

จาก RCA: FN หนักสุดคือ **sliver/residue GT** (คราบบาง, เศษจิ๋ว, polygon aspect เพี้ยน) —
23% ของ test GT. ใช้กฎมาตรฐานจาก literature:

- **strict** — eval เดิม
- **sieved** — ตัด GT+pred ที่ area<0.5% หรือ aspect>8:1 (aerial-imagery sieving)
- **void** — sliver GT = void: ไม่นับ FN; pred overlap void>50% ไม่นับ FP (Kirillov CVPR19)
- **kirillov** — void เต็มรูปแบบ: slivers + *group label* (instances ติดกันแยกไม่ได้)
- **biou** — strict matching แต่ score ด้วย Boundary IoU (Cheng CVPR21)

แล้วเทรน **arm_clean960** = gold_aug_clean (sliver GT ถูกตัดจาก train/valid) @ imgsz 960 —
ตรงเทคนิค "small-instance-free training" (arXiv 2412.05611) + ความละเอียดสูง


```python
# ============================================================
# 9.1 DATASET SURGERY -> data/gold_aug_clean
# drop sliver/residue GT (area<0.5% img OR bbox aspect>8:1) from
# train+valid labels; test labels byte-identical (protocols at eval).
# images are hardlinked (same volume) to avoid duplicating ~200MB.
# ============================================================
SIEVE_AREA, SIEVE_ASP = 0.005, 8.0
SRC_AUG = DATASETS["gold_aug"]
DST_CLN = DATA_DIR / "gold_aug_clean"

def _poly_area_norm(xy):
    x, y = xy[0::2], xy[1::2]
    return 0.5 * abs(float(np.dot(x, np.roll(y, -1))
                         - np.dot(y, np.roll(x, -1))))

def _keep_line(ln, W, H):
    p = ln.split()
    if len(p) < 7:
        return False
    xy = np.array(p[1:], float)
    if _poly_area_norm(xy) < SIEVE_AREA:
        return False
    px, py = xy[0::2] * W, xy[1::2] * H
    bw, bh = px.max() - px.min(), py.max() - py.min()
    return max(bw, bh) / max(1.0, min(bw, bh)) <= SIEVE_ASP

surg = {}
for sp in ("train", "valid", "test"):
    (DST_CLN / sp / "images").mkdir(parents=True, exist_ok=True)
    (DST_CLN / sp / "labels").mkdir(parents=True, exist_ok=True)
    kept = dropped = 0
    for ip in sorted((SRC_AUG / sp / "images").iterdir()):
        dp = DST_CLN / sp / "images" / ip.name
        if not dp.exists():
            try:
                os.link(ip, dp)
            except OSError:
                shutil.copy2(ip, dp)
        lf = SRC_AUG / sp / "labels" / (ip.stem + ".txt")
        out = DST_CLN / sp / "labels" / (ip.stem + ".txt")
        if out.exists():
            continue
        if sp == "test" or not lf.exists():
            out.write_text(lf.read_text() if lf.exists() else "")
            continue
        with Image.open(ip) as im:
            W, H = im.size
        lines = []
        for ln in lf.read_text().strip().splitlines():
            if _keep_line(ln, W, H):
                lines.append(ln); kept += 1
            else:
                dropped += 1
        out.write_text("\n".join(lines))
    surg[sp] = (kept, dropped)
    print(f"{sp:6s}: kept={kept} dropped_slivers={dropped}")

CLEAN_YAML = CFG_DIR / "gold_aug_clean.yaml"
CLEAN_YAML.write_text(yaml.safe_dump(
    {"path": str(DST_CLN.resolve()),
     "train": "train/images", "val": "valid/images", "test": "test/images",
     "nc": 2, "names": ["gravel", "sand"]}))
DATASETS["gold_aug_clean"] = DST_CLN
DATA_YAML["gold_aug_clean"] = str(CLEAN_YAML)
print("yaml ->", CLEAN_YAML, "| ds_sig:", ds_sig(CLEAN_YAML))
```

```python
# ============================================================
# 9.2 CONVENTION-AWARE EVAL HARNESS + baseline re-eval
# protocols: strict | sieved | void | kirillov(void+group) | biou
# COCO-style mask AP50-95 (custom, verified ~= ultralytics) + FP/FN
# ============================================================
import cv2

SIEVE_AREA, SIEVE_ASP = 0.005, 8.0   # same rule as 9.1 surgery
IOU_THRS = np.round(np.arange(0.5, 1.0, 0.05), 2)
CONF_MATCH = 0.25
BIOU_D = 0.02          # Cheng default: 2% of image diagonal
GRP_D  = 0.01          # group merge: masks touching after 1%-diag dilation

def _sieve(g):
    return g["area"] >= SIEVE_AREA and g["aspect"] <= SIEVE_ASP

def _miou(a, b):
    u = (a | b).sum()
    return (a & b).sum() / u if u else 0.0

def _band(m, d):
    m8 = m.astype(np.uint8)
    return ((cv2.distanceTransform(m8, cv2.DIST_L2, 3) <= d)
            | (cv2.distanceTransform(1 - m8, cv2.DIST_L2, 3) <= d))

def _biou(a, b, d):
    ba, bb = _band(a, d), _band(b, d)
    u = (ba | bb).sum()
    return float((ba & bb).sum() / u) if u else 1.0

def _load_gt(lf, W, H):
    out = []
    if not lf.exists():
        return out
    for ln in lf.read_text().strip().splitlines():
        p = ln.split()
        if len(p) < 7:
            continue
        xy = np.array(p[1:], float).reshape(-1, 2)
        px = (xy * [W, H]).astype(np.int32)
        m = np.zeros((H, W), np.uint8)
        cv2.fillPoly(m, [px], 1)
        bw, bh = np.ptp(px[:, 0]), np.ptp(px[:, 1])
        out.append(dict(cls=int(p[0]), mask=m.astype(bool),
                        area=float(m.mean()),
                        aspect=float(max(bw, bh) / max(1, min(bw, bh)))))
    return out

def load_split(ckpt, split, imgsz, conf=0.001):
    """predict every image once; return GT[name]->list, PR[name]->list."""
    model = YOLO(str(ckpt))
    root = Path(yaml.safe_load(Path(DATA_YAML["gold"]).read_text())["path"])
    imgdir = root / split / "images"; lbldir = root / split / "labels"
    GT, PR = {}, {}
    for k, ip in enumerate(sorted(imgdir.glob("*"))):
        im = cv2.imread(str(ip))
        if im is None: continue
        H, W = im.shape[:2]
        GT[ip.name] = _load_gt(lbldir / (ip.stem + ".txt"), W, H)
        r = model.predict(str(ip), imgsz=imgsz, conf=conf, device=DEVICE,
                          retina_masks=True, verbose=False)[0]
        pr = []
        if r.masks is not None:
            for c, cf, m in zip(r.boxes.cls.cpu().numpy().astype(int),
                                r.boxes.conf.cpu().numpy(),
                                r.masks.data.cpu().numpy()):
                mm = cv2.resize(m, (W, H)) > 0.5
                if not mm.any(): continue
                ys, xs = np.where(mm)
                bw, bh = np.ptp(xs), np.ptp(ys)
                pr.append(dict(cls=int(c), conf=float(cf), mask=mm,
                               area=float(mm.mean()),
                               aspect=float(max(bw, bh) / max(1, min(bw, bh)))))
        PR[ip.name] = pr
        if (k + 1) % 25 == 0:
            print(f"  {k+1}/{len(list(imgdir.glob('*')))}", flush=True)
    return GT, PR, (W ** 2 + H ** 2) ** 0.5

def _clusters(gt_c, d):
    """union-find: same-class GT whose d-dilated masks touch."""
    n = len(gt_c); par = list(range(n))
    def find(x):
        while par[x] != x:
            par[x] = par[par[x]]; x = par[x]
        return x
    for i in range(n):
        di = cv2.dilate(gt_c[i]["mask"].astype(np.uint8),
                        np.ones((max(3, int(d)),) * 2, np.uint8)).astype(bool)
        for j in range(i + 1, n):
            if (di & gt_c[j]["mask"]).any():
                pi, pj = find(i), find(j)
                if pi != pj: par[pi] = pj
    sz = {}
    for i in range(n): sz[find(i)] = sz.get(find(i), 0) + 1
    return {i for i in range(n) if sz[find(i)] > 1}

def _kv(gt_c, proto, d):
    """-> (keep, void) GT lists for one image+class."""
    if proto in ("strict", "biou"):
        return gt_c, []
    keep = [g for g in gt_c if _sieve(g)]
    sl   = [g for g in gt_c if not _sieve(g)]
    if proto == "sieved":
        return keep, []
    if proto == "void":
        return keep, sl
    if proto == "kirillov":
        grp = _clusters(keep, GRP_D * d)
        return ([g for i, g in enumerate(keep) if i not in grp],
                sl + [g for i, g in enumerate(keep) if i in grp])
    raise ValueError(proto)

def _pred_ok(p, proto):
    return proto != "sieved" or _sieve(p)

def ap_class(GT, PR, diag, proto, cls, biou=False):
    score = (lambda a, b: _biou(a, b, BIOU_D * diag)) if biou else _miou
    n_gt = sum(1 for im in GT.values() for g in im
               if g["cls"] == cls and len(_kv([g], proto, diag)[0]) == 1)
    # NOTE: for kirillov, keep-set depends on whole image cluster -> recompute below
    aps = []
    for thr in IOU_THRS:
        recs = []
        n_gt = 0
        for nm in GT:
            gt_c = [g for g in GT[nm] if g["cls"] == cls]
            keep, void = _kv(gt_c, proto, diag)
            n_gt += len(keep)
            used = set()
            for p in sorted([p for p in PR[nm] if p["cls"] == cls
                             and _pred_ok(p, proto)], key=lambda x: -x["conf"]):
                best, bj = 0.0, -1
                for j, g in enumerate(keep):
                    if j in used: continue
                    v = score(p["mask"], g["mask"])
                    if v > best: best, bj = v, j
                if best >= thr:
                    used.add(bj); recs.append((p["conf"], 1))
                else:
                    vv = max([(p["mask"] & g["mask"]).sum() / p["mask"].sum()
                              for g in void] + [0.0])
                    if vv < 0.5:
                        recs.append((p["conf"], 0))
        if n_gt == 0:
            aps.append(np.nan); continue
        recs.sort(key=lambda x: -x[0])
        tp = np.cumsum([r[1] for r in recs]); fp = np.cumsum([1 - r[1] for r in recs])
        rec = tp / n_gt; prec = tp / np.maximum(tp + fp, 1)
        mrec = np.concatenate([[0], rec, [1]]); mpre = np.concatenate([[0], prec, [0]])
        for i in range(len(mpre) - 2, -1, -1):
            mpre[i] = max(mpre[i], mpre[i + 1])
        idx = np.searchsorted(mrec, np.linspace(0, 1, 101), side="left")
        aps.append(mpre[np.minimum(idx, len(mpre) - 1)].mean())
    return float(np.nanmean(aps))

def fpfn_class(GT, PR, diag, proto, cls, conf=CONF_MATCH):
    tp = fp = fn = 0
    for nm in GT:
        gt_c = [g for g in GT[nm] if g["cls"] == cls]
        keep, void = _kv(gt_c, proto, diag)
        used = set()
        for p in sorted([p for p in PR[nm] if p["cls"] == cls
                         and p["conf"] >= conf and _pred_ok(p, proto)],
                        key=lambda x: -x["conf"]):
            best, bj = 0.0, -1
            for j, g in enumerate(keep):
                if j in used: continue
                v = _miou(p["mask"], g["mask"])
                if v > best: best, bj = v, j
            if best >= 0.5:
                used.add(bj); tp += 1
            else:
                vv = max([(p["mask"] & g["mask"]).sum() / p["mask"].sum()
                          for g in void] + [0.0])
                if vv < 0.5: fp += 1
        fn += len(keep) - len(used)
    return tp, fp, fn

PROTOS = ["strict", "sieved", "void", "kirillov", "biou"]

def conv_eval_table(ckpt, split="test", imgsz=640, tag="model"):
    GT, PR, diag = load_split(ckpt, split, imgsz)
    rows = []
    for proto in PROTOS:
        biou = proto == "biou"
        base = "strict" if biou else proto
        apg = ap_class(GT, PR, diag, base, 0, biou=biou)
        aps = ap_class(GT, PR, diag, base, 1, biou=biou)
        tg, fg, ng = fpfn_class(GT, PR, diag, base, 0)
        ts, fs, ns = fpfn_class(GT, PR, diag, base, 1)
        rows.append(dict(model=tag, protocol=proto,
                         AP_gravel=round(apg, 4), AP_sand=round(aps, 4),
                         mAP5095=round(np.nanmean([apg, aps]), 4),
                         TP=tg + ts, FP=fg + fs, FN=ng + ns))
        print(f"{tag:10s} {proto:8s} mAP={rows[-1]['mAP5095']} "
              f"TP={rows[-1]['TP']} FP={rows[-1]['FP']} FN={rows[-1]['FN']}")
    return pd.DataFrame(rows)

BASELINE_CKPT = sorted(RUNS_DIR.glob("arm_gold-2*/weights/best.pt"))[-1]
print("=== baseline arm_gold-2, test split ===")
conv_base = conv_eval_table(BASELINE_CKPT, "test", 640, "baseline")
conv_base.to_csv(RES_DIR / "convention_eval_baseline.csv", index=False)
conv_base
```
